# Generate Student Performance Dataset
This notebook generate a synthetic student_performance dataset for practicing data validation, cleaning, analysis, and visualization.
The generate dataset is saved to: `data/raw/student_performance.csv`

```python
from pathlib import Path 
import numpy as np
import pandas as pd
```

```python
PROJECT_ROOT = Path("..").resolve()

RAW_DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "student_performance.csv"
)

RAW_DATA_PATH.parent.mkdir(
    parents=True,
    exist_ok=True,
)

print("Output path:", RAW_DATA_PATH)
```

```python
rng = np.random.default_rng(42)
number_of_students = 200

study_hours = np.clip(
    rng.normal(
        loc=10,
        scale=4,
        size=number_of_students,
    ),
    0,
    30,
)

attendance_rate = np.clip(
    rng.normal(
        loc=80,
        scale=12,
        size=number_of_students,
    ),
    40,
    100,
)

base_score = (
    1.5
    + 0.25 * study_hours
    + 0.04 * attendance_rate
    + rng.normal(
        loc=0,
        scale=1.2,
        size=number_of_students,
    )
)

student_df = pd.DataFrame({
    "student_id": [
        f"STU{i:03d}"
        for i in range(1, number_of_students + 1)
    ],
    "gender": rng.choice(
        ["Female", "Male", "Other"],
        size=number_of_students,
    ),
    "major": rng.choice(
        [
            "Software Engineering",
            "Computer Science",
            "Information Systems",
        ],
        size=number_of_students,
    ),
    "study_hours": np.round(study_hours, 1),
    "attendance_rate": np.round(attendance_rate, 1),
    "assignment_score": np.round(
        np.clip(
            base_score + rng.normal(0, 0.8, number_of_students),
            0,
            10,
        ),
        1,
    ),
    "midterm_score": np.round(
        np.clip(
            base_score + rng.normal(0, 1.0, number_of_students),
            0,
            10,
        ),
        1,
    ),
    "final_score": np.round(
        np.clip(
            base_score + rng.normal(0, 1.1, number_of_students),
            0,
            10,
        ),
        1,
    ),
})
```

```python
student_df.loc[5, "assignment_score"] = np.nan
student_df.loc[12, "attendance_rate"] = 105
student_df.loc[20, "major"] = " computer science "
student_df.loc[30, "final_score"] = -1
student_df.loc[45, "study_hours"] = -3
student_df.loc[60, "gender"] = " female "

student_df = pd.concat(
    [
        student_df,
        student_df.iloc[[0]],
    ],
    ignore_index=True,
)
```

```python
student_df.to_csv(
    RAW_DATA_PATH,
    index = False,
)

print("Dataset saved successfully.")
print("Shape:", student_df.shape)
print("File size:", RAW_DATA_PATH.stat().st_size, "bytes")
```
